```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Walmart_Sales.csv")
df.head()
```

```python
sales_by_store = df.groupby("Store")["Weekly_Sales"].sum()
```

```python
plt.figure(figsize=(10,10))
plt.pie(
    sales_by_store,
    labels=sales_by_store.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Market Share of Weekly Sales by Store")
plt.axis("equal") 
plt.show()
```
