### Task 2: Important Mathmetical Operation

```python
A = np.array([10,20,30,40])
B = np.array([1,2,3,4])
added = A + B
print(added)
subtracted = A - B
print(subtracted)
multiplied = A * B
print(multiplied)
divided = A / B
print(divided)
power = A ** 2
print(power)
#-------By Using Numpy Functions-------
print("Using Numpy Functions:")
added_func = np.add(A, B)
print(added_func)
subtracted_func = np.subtract(A, B)
print(subtracted_func)
multiplied_func = np.multiply(A, B)
print(multiplied_func)
divided_func = np.divide(A, B)
print(divided_func)
power_func = np.power(A, 2)
print(power_func)
```

### Task 3: Important Numpy Mathmatical Function

```python
values = np.array([2,4,6,8,10])
sqrt_values = np.sqrt(values)
print("Square Root:", sqrt_values)
exp_values = np.exp(values)
print("Exponential:", exp_values)
log_values = np.log(values)
print("Logarithm:", log_values)
sum_values = np.sum(values)
print("Sum:", sum_values)
cumsum_values = np.cumsum(values)
print("Cumulative Sum:", cumsum_values)
```

### Task 4: Aggreation Operation 

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
row_sum1 = np.sum(data, axis=1)
print("Row-wise Sum:", row_sum1)
col_sum1 = np.sum(data, axis=0)
print("Column-wise Sum:", col_sum1)
min_value = np.min(data)
print("Minimum Value:", min_value)
max_value = np.max(data)
print("Maximum Value:", max_value)
mean_value = np.mean(data)  
print("Overall Mean:", mean_value)
```

### Task 5: Statistical Oprations

```python
marks = np.array([78,85,90,66,72,88,95,60])
mean_marks = np.mean(marks)
print("Mean Marks:", mean_marks)
median_marks = np.median(marks)
print("Median Marks:", median_marks)
variance_marks = np.var(marks)
print("Variance of Marks:", variance_marks)
std_dev_marks = np.std(marks)
print("Standard Deviation of Marks:", std_dev_marks)
minimum_marks = np.min(marks)
print("Minimum Marks:", minimum_marks)
maximum_marks = np.max(marks)
print("Maximum Marks:", maximum_marks)
range_marks = maximum_marks - minimum_marks
print("Range of Marks:", range_marks)
```

### Task 6: Percentiles and Sorting

```python
sorted_marks = np.sort(marks)
print("Sorted Marks:", sorted_marks)
percentiles = np.percentile(marks, [25, 50, 75])
print("25th, 50th, 75th Percentiles:", percentiles)
avarage_marks = np.mean(marks)
print("Average Marks:", avarage_marks)
above_average_count = np.sum(marks > avarage_marks)
print("Count of Marks Above Average:", above_average_count)
```

### Task 7: Mini Use Case (Sales Analysis)

```python
sales = np.array([1200,1500,900,2000,1800,1700,1600])
total_sales = np.sum(sales)
print("Total Weekly Sales:", total_sales)
average_sales = np.mean(sales)
print("Average Daily Sales:", average_sales)
highest_sales = np.max(sales)
print("Highest Sales in a Day:", highest_sales)
lowest_sales = np.min(sales)
print("Lowest Sales in a Day:", lowest_sales)
std_dev_sales = np.std(sales)
print("Standard Deviation of Sales:", std_dev_sales)
above_average_sales = np.sum(sales > average_sales)
print("Number of Days with Sales Above Average:", above_average_sales)
```
