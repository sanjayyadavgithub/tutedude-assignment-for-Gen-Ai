### Task 1: Basic Class and Object Creation

```python
class Product():
    def __init__(self,name,price,category):
        self.name = name
        self.price = price
        self.category = category
    def get_info(self):
        return f"Product Name: {self.name}, Price: {self.price}, Category: {self.category}"
    def apply_discount(self, percentage):
        discounted_price = self.price * (1 - percentage / 100)
        return discounted_price
product1 = Product("Laptop", 1000, "Electronics")
product2 = Product("Shirt", 50, "Clothing")
print("Product Information 1 : ")
print(product1.get_info())
discounted_price = product1.apply_discount(10)
print(f"Discounted Price of {product1.name} after applying 10% discount: {discounted_price}")
print("Product Information 2 : ")
print(product2.get_info())
discounted_price = product2.apply_discount(20)
print(f"Discounted Price of {product2.name} after applying 20% discount: {discounted_price}")
```

### Task 2: Constructor and Encapsulation

```python
class Product():
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price  
        self.category = category
    def get_info(self):
        return f"Product Name: {self.name}, Price: {self.__price}, Category: {self.category}"
    def apply_discount(self, percentage):
        discounted_price = self.__price * (1 - percentage / 100)
        return discounted_price
    def get_price(self):
        return self.__price
    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Price must be positive")

product1 = Product("Laptop", 1000, "Electronics")
product2 = Product("Shirt", 50, "Clothing")
print("Product Information 1 : ")
print(product1.get_info())
discounted_price = product1.apply_discount(10)
print(f"Discounted Price of {product1.name} after applying 10% discount: {discounted_price}")
print("Product Information 2 : ")
print(product2.get_info())
discounted_price = product2.apply_discount(20)
print(f"Discounted Price of {product2.name} after applying 20% discount: {discounted_price}")
print(f"\nGetting price using getter: {product1.get_price()}")
product1.set_price(1200)
print(f"Updated price: {product1.get_price()}")
```

### Task 3: Inheritrance (Single Level)

```python
class ElectronicsProduct(Product):
    def __init__(self, name, price, category, warranty_years):
        super().__init__(name, price, category)
        self.warranty_years = warranty_years
    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, Warranty Period: {self.warranty_years} years"

product1 = ElectronicsProduct("Laptop", 1000, "Electronics", 2)
print("Electronics Product Information : ")
print(product1.get_info())
```

### Task 4: Polymorphism

```python
class Laptop(Product):
    def __init__(self, name, price, category, brand):
        super().__init__(name, price, category)
        self.brand = brand
    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, Brand: {self.brand}"
class Mobile(Product):
    def __init__(self, name, price, category, brand):
        super().__init__(name, price, category)
        self.brand = brand
    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, Brand: {self.brand}"
laptop1 = Laptop("Gaming Laptop", 1500, "Electronics", "BrandA")
mobile1 = Mobile("Smartphone", 800, "Electronics", "BrandB")
print("Laptop Information : ")
print(laptop1.get_info())
print("Mobile Information : ")
print(mobile1.get_info())

```

### Task 5: Abstraction Using Abstract Base Class

```python
#Abstraction using abstract base class with abstract methods
class PaymentMethod():
    def process_payment(self, amount):
        pass
class CreditCardPayment(PaymentMethod):
    def process_payment(self, amount):
        return f"Processing credit card payment of ${amount}"
class UPIPayment(PaymentMethod):
    def process_payment(self, amount):
        return f"Processing UPI payment of ${amount}"   
credit_payment = CreditCardPayment()
upi_payment = UPIPayment()
print(credit_payment.process_payment(100))
print(upi_payment.process_payment(200))
```

### Task 6: Magic Method and Operator Overloading

```python
class ProductMagic():
    def __init__(self,name,price,category):
        self.name = name
        self.price = price
        self.category = category
    def get_info(self):
        return f"Product Name: {self.name}, Price: {self.price}, Category: {self.category}"
    def apply_discount(self, percentage):
        discounted_price = self.price * (1 - percentage / 100)
        return discounted_price
    def __str__(self):
        return f"Product({self.name}, {self.price}, {self.category})"

    def __add__(self, other):
        if isinstance(other, ProductMagic):
            return self.price + other.price
        return NotImplemented

product1 = ProductMagic("Laptop", 1000, "Electronics")
product2 = ProductMagic("Shirt", 50, "Clothing")
print(product1)
print(product2)
print("Sum of prices of product1 and product2: ", product1 + product2)
```

### Task 7: Mini Project (Simple Inventry System)

```python
class Inventry():
    def __init__(self):
        self.products = []
    def add_product(self, product):
        self.products.append(product)
    def remove_product(self, product):
        self.products.remove(product)
    def get_total_value(self):
        return sum(product.price for product in self.products)
    def show_all_products(self):
        for product in self.products:
            print(product.get_info())
class Store():
    def __init__(self, store_name):
        self.store_name = store_name
        self.inventry = Inventry()
    def add_product_to_store(self, product):
        self.inventry.add_product(product)
    def remove_product_from_store(self, product):
        self.inventry.remove_product(product)
    def get_store_value(self):
        return self.inventry.get_total_value()
    def show_store_products(self):
        print(f"Products in {self.store_name}:")
        self.inventry.show_all_products()
store = Store("My Store")
product1 = ProductMagic("Laptop", 1000, "Electronics")
product2 = ProductMagic("Shirt", 50, "Clothing")
product3 = ProductMagic("Phone", 500, "Electronics")
product4 = ProductMagic("Jeans", 80, "Clothing")
product5 = ProductMagic("Headphones", 150, "Electronics")
store.add_product_to_store(product1)
store.add_product_to_store(product2)
store.add_product_to_store(product3)
store.add_product_to_store(product4)
store.add_product_to_store(product5)
print("Store Information : ")
store.show_store_products()
print(f"Total value of products in store: ${store.get_store_value()}")
print("\nRemoving product2 (Shirt) from store...")
store.remove_product_from_store(product2)
store.show_store_products()
print(f"Total value of products in store after removal: ${store.get_store_value()}")
print("\nadd two product price using operator overloading: ")
print(f"Sum of prices of product1 and product3: {product1 + product3}")

```
