### Task 1: Pandas Series Basic


```python
import pandas as pd
marks = [78,85,90,66,72]
series_marks = pd.Series(marks)
print(series_marks)
index_marks = pd.Index(marks)
print(index_marks)
data_type_marks = series_marks.dtype
print("Data Type of Series:", data_type_marks)
first_mark = series_marks[0]
print("First Mark:", first_mark)
last_two_marks = series_marks[-2:]
print("Last Two Marks:\n", last_two_marks)
```

### Task 2: Mathmatical Operation On Series

```python
# add 5 grace points to each mark
grace_marks = series_marks + 5
print("Marks after adding grace points:\n", grace_marks)
subtracted_marks = series_marks - 2
print("Marks after subtracting 2 points:\n", subtracted_marks)
multiplied_marks = series_marks * 1.05
print("Marks after multiplying by 1.05:\n", multiplied_marks)
divided_marks = series_marks / 2
print("Marks after dividing by 2:\n", divided_marks)
```

### Task 3: Python Functionalty on Series


```python
max_mark = series_marks.max()
print("Maximum Mark:", max_mark)
min_mark = series_marks.min()
print("Minimum Mark:", min_mark)
sum_marks = series_marks.sum()
print("Sum of Marks:", sum_marks)
mean_mark = series_marks.mean()
print("Mean Mark:", mean_mark)
pass_mark = series_marks.apply(lambda x: x >= 70)
print("Marks above 70:\n", pass_mark)
count_pass_mark = pass_mark.sum()
print("Count of Marks above 70:", count_pass_mark)
```

### Task 4: Create A DataFrame

```python
# create a DataFrame
students = {
    "Name" : ['Amit','Neha','Rahul','Sneha','Pooja'],
    "Marks" : [78,85,90,66,72],
    "Subject" : ['Math','Math','Science','Science','Math']
}
df_students = pd.DataFrame(students)
print("Data Frame : ",df_students)
print("First 3 rows :\n", df_students.head(3))
print("Last 2 rows :\n", df_students.tail(2))
print("Shape of Data Frame : ", df_students.shape)
print("Columns of Data Frame : ", df_students.columns)
print("Data Types of Data Frame :\n", df_students.dtypes)
```

### Task 5: Important data Frame Function

```python
print("Student data frame info :\n", df_students.info())
print("Summary Statistics :\n", df_students.describe())
print("Head of data From :\n", df_students.head())
print("tail of data From :\n", df_students.tail())
sorted_students = df_students.sort_values(by='Marks', ascending=False)
print("Students sorted by Marks (Descending):\n", sorted_students)
sorted_students.reset_index(drop=True, inplace=True)
print("Sorted Students with Reset Index:\n", sorted_students)
```

### Task 6: Filtering and Conditional Selection

```python
above_75_students = df_students[df_students['Marks'] > 75]
print("Students with Marks above 75:\n", above_75_students)
math_students = df_students[df_students['Subject'] == 'Math']
print("Students belonging to Math subject:\n", math_students)
average_marks = df_students['Marks'].mean()
above_average_students = df_students[df_students['Marks'] > average_marks]
print("Students with Marks above Average:\n", above_average_students)
failed_students = df_students[df_students['Marks'] < 70]
print("Students who failed (Marks below 70):\n", failed_students)

```

### Task 7: Grouping and Basic Analysis

```python
average_marks_per_subject = df_students.groupby('Subject')['Marks'].mean()
print("Average Marks per Subject:\n", average_marks_per_subject)
students_count_per_subject = df_students.groupby('Subject')['Name'].count()
print("Count of Students per Subject:\n", students_count_per_subject)
max_marks_per_subject = df_students.groupby('Subject')['Marks'].max()
print("Maximum Marks per Subject:\n", max_marks_per_subject)
```

### Task 8: Pandas Ploting (Simple Graph)

```python
df_students.plot.bar(x='Name', y='Marks', title='Distribution of Marks')
```

```python
df_students.plot.line(x='Name', y='Marks', title='Marks Line Plot')
```

```python
df_students.plot.hist(y='Marks', bins=5, title='Marks Distribution')
```

### Task 9: Mini Use Case (Sales Data Analysis)

```python
salesData = {
    'Day':['Mon','Tue','Wed','Thu','Fri'],
    'Revenue':[1200, 1500, 900, 2000, 1800]
}
total_revenue = sum(salesData['Revenue'])
print("Total Revenue:", total_revenue)
average_revenue = total_revenue / len(salesData['Revenue'])
print("Average Daily Revenue:", average_revenue)
max_revenue = max(salesData['Revenue'])
max_revenue_index = salesData['Revenue'].index(max_revenue)
day_with_max_revenue = salesData['Day'][max_revenue_index]
print("Day with Highest Revenue:", day_with_max_revenue)
days_above_average = [salesData['Day'][i] for i in range(len(salesData['Revenue'])) if salesData['Revenue'][i] > average_revenue]
print("Days with Revenue above Average:", days_above_average)
sales_df = pd.DataFrame(salesData)
sales_df.plot.bar(x='Day', y='Revenue', title='Daily Revenue')
sales_df.plot.line(x='Day', y='Revenue', title='Daily Revenue Trend')
sales_df.plot.hist(y='Revenue', bins=5, title='Revenue Distribution')
```
