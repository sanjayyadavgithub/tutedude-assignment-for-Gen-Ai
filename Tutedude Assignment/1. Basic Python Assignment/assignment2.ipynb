### Task 1 : Discount Rules 

```python
order_amount = input("Enter Order Amount: ")
amount = 0 
tax = 0
try:
    order_amount = int(order_amount)
    if order_amount >= 2000:
        amount = order_amount - (order_amount * 0.15)
        tax = order_amount * 0.15
    elif order_amount >= 1500 and order_amount < 2000:
        amount = order_amount - (order_amount * 0.10)
        tax = order_amount * 0.10
    elif order_amount >= 1000 and order_amount < 1500:
        amount = order_amount - (order_amount * 0.07)
        tax = order_amount * 0.07
    else:
        amount = order_amount- (order_amount * 0.05)
        tax = order_amount * 0.05
    print("Final Amount to be paid: ", amount , "Tax: ", tax)
    amount = amount + (amount * 0.05)
    tax = amount * 0.05
    print("Final Amount to be paid with GST: ", amount, "GST: ", tax)
except:
    print("Invalid input. Please enter a numeric value for the order amount.")

```

### Task 2: Process Multiple Orders


```python
order_amount_list = [1200, 2500 , 800, 1750, 3000]
total_revenue = 0
number_of_orders = 0
for order_amount in order_amount_list:
    amount = 0 
    tax = 0
    if order_amount >= 2000:
        amount = order_amount - (order_amount * 0.15)
        tax = order_amount * 0.15
    elif order_amount >= 1500 and order_amount < 2000:
        amount = order_amount - (order_amount * 0.10)
        tax = order_amount * 0.10
    elif order_amount >= 1000 and order_amount < 1500:
        amount = order_amount - (order_amount * 0.07)
        tax = order_amount * 0.07
    else:
        amount = order_amount- (order_amount * 0.05)
        tax = order_amount * 0.05
    total_revenue += amount
    if(order_amount >= amount):
        number_of_orders += 1

    print("Final Amount to be paid: ", order_amount, "Tax: ", tax ,"total: ", amount)
    amount = amount + (amount * 0.05)
    tax = amount * 0.05
    print("Final Amount to be paid with GST: " , order_amount, "Tax: ", tax ,"total: ", amount)
print("Total Revenue: ", total_revenue)
print("Number of Orders: ", number_of_orders)
```

### Task 3: User Menu (While loop + break/continue)

```python
orders = []
while True:
    print("\nMenu:")
    print("1 - Add order amount")
    print("2 - Show all orders and totals after discounts")
    print("q - Quit")
    choice = input("Enter your choice: ")
    if choice == "q":
        print("Exiting program...")
        break
    elif choice == "1":
        amount = float(input("Enter order amount: "))
        orders.append(amount)
        print("Order added successfully.")
    elif choice == "2":
        if len(orders) == 0:
            print("No orders entered yet.")
        else:
            total_before = 0
            total_after = 0
            print("\nOrders Summary:")
            for order in orders:
                total_before += order
                if order > 100:
                    discounted = order * 0.90
                else:
                    discounted = order

                total_after += discounted
                print("Original:", order, " Discounted:", discounted)

            print("Total before discount:", total_before)
            print("Total after discount:", total_after)

    else:
        print("Invalid input. Please try again.")
        continue
```

### Task 4: Loop Controll With Condition (Break/Continue)

```python
daily = [200, 150, 0, 400, 50, -1, 300]
total_sales = 0
for sale in daily:
    if sale == -1:
        print("Corrupted data found. Stopping...")
        break
    if sale == 0:
        print("No sales today. Skipping...")
        continue
    total_sales += sale
    print("Running total:", total_sales)

print("Final total sales:", total_sales)
```
