### Task 3 : Product Pricing (Dictionaries)

```python
price_dict = [
    ("Laptop", 75000),
    ("Mobile", 30000),
    ("Headphones", 2000),
    ("Keyboard", 1500),
    ("Mouse", 800),
    ("Monitor", 12000)
]
price_dict = dict(price_dict)
print("Price of Laptop:", price_dict["Laptop"])
price_dict["Tablet"] = 25000
print("Updated price dictionary:", price_dict)
del_key = "Mobile"
if del_key in price_dict:
    del price_dict[del_key]
    print(f"'{del_key}' removed from price dictionary.")
else:    print(f"'{del_key}' not found in price dictionary.")
print("Price dictionary after deletion:", price_dict)
price = price_dict.values()
print("All prices:", price)
sun_of_products = sum(price)
print("Sum of all product prices:", sun_of_products)
lenth_of_products = len(price_dict)
print("Number of products:", lenth_of_products)
avarage_price = sun_of_products / lenth_of_products
print("Average price of products:", avarage_price)
max_product = max(price_dict.items(), key=lambda x: x[1])
print("Product with the highest price:", max_product)
min_product = min(price_dict.items(), key=lambda x: x[1])
print("Product with the lowest price:", min_product)
```

### Task 4 : Combined Operation

```python
products = ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse", "Monitor"]
price_dict = {
    "Laptop": 75000,
    "Mobile": 30000,
    "Headphones": 2000,
    "Keyboard": 1500,
    "Mouse": 800,
    "Monitor": 12000
}
categories = ["Electronics", "Electronics", "Accessories", "Accessories", "Accessories", "Electronics"]
catalog = []
for i in range(len(products)):
    name = products[i]
    price = price_dict[name]
    category = categories[i]
    catalog.append((name, price, category))

print("Catalog:", catalog)

category_to_products = {}
for name, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []
    category_to_products[category].append(name)

print("Category Mapping:", category_to_products)
max_category = max(category_to_products, key=lambda x: len(category_to_products[x]))
print("Category with max products:", max_category)
print("Products:", category_to_products[max_category])
```
