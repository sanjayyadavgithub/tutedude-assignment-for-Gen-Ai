### Task 1: Safe Division Utility

```python
try:
    numerator = float(input("Enter the numerator: "))
    denominator = float(input("Enter the denominator: "))
    result = numerator / denominator
    print("The result of division is: ", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero. Please provide a non-zero denominator.")
except ValueError:  
    print("Error: Invalid input. Please enter numeric values for numerator and denominator.")
finally:
    print("Division operation completed.")
```

### Task 2: Billing Calculator with Error Handling

```python
prices = [120,350,'abc', 500 ,-200, 800]
total_price = 0
for price in prices:
    try:
        if price < 0:
            raise ValueError("Price cannot be negative.")
        total_price += price
    except ValueError as ve:
        print(f"Error: {ve} - Invalid price: {price}")
    except TypeError:
        print(f"Error: Invalid type for price: {price}. Price must be a number.")
    finally:
        print("Processed price :  ",price)
print("Total price of valid items: ", total_price)
```

### Task 3: Custom Exception Age validation

```python
def check_age(age):
    if age <= 0:
        raise ValueError("Age cannot be negative.")
    elif age > 0 and age < 120:
        print("Access granted. Welcome!")
    else:
        raise ValueError("Age must be between 1 to 120.")
try:
    user_age = int(input("Enter your age: "))
    check_age(user_age)
except ValueError as ve:
    print(f"Error: {ve}")
    
```

### Task 4: File Reader with Exception Handler

```python
file_name = input("Enter the filename to read: ")
constent = ""
try:
    with open(file_name, 'r') as file:
        content = file.read()
        constent = content
        for line in constent.splitlines():
            if constent.splitlines().index(line) < 3:
                print(line)
except FileNotFoundError:
    print(f"Error: The file '{file_name}' was not found. Please check the filename and try again.")
except PermissionError:
    print(f"Error: You do not have permission to read the file '{file_name}'. Please check your permissions and try again.")
finally:
    print("File operation attempted.")
   
```

### Task 5: Mini Program: Safe Shopping Cart

```python
def isNumber(num):
    if not isinstance(num, (int, float)):
        return False
    return True
cart = []
while True:
    item = input("Enter an item to add to the cart (or type 'q' to finish): ")
    if item.lower() == 'q':
        break
    try:
        if isNumber(item):
            raise ValueError("Item name must be a string without numbers or special characters.")
        item = float(item) 
        if item < 0:
            raise ValueError("Item price cannot be negative.")
        cart.append(item)
    except ValueError as ve:
        print(f"Error: {ve} - Invalid item: {item}")
    finally:
        print("Processed item: ", item)
print("Items in cart: ", cart)
print("Total items in cart: ", len(cart))
print("Total price of items in cart: ", sum(cart))
```
