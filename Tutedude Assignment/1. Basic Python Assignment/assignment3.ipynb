### Task 1: Functions Basic 

```python
def apply_discount(price, discount_percent=0):
    if discount_percent < 60:
        if discount_percent > 0 :
            discount_amount = price * (discount_percent / 100)
            discounted_price = price - discount_amount
            return discounted_price
        else:
            discount_amount = price * (5 / 100)
            discounted_price = price - discount_amount
            return discounted_price
    else:
        print("Discount too high, cannot apply.")
        return price
# Example usage
print(apply_discount(100, 10))  # Output: 90.0
print(apply_discount(100, 6))   # Output: 95.0
print(apply_discount(100))  # Output: 95.0
print(apply_discount(100, 70))  # Output: Discount too high, cannot apply. 100
```

### Task 2: Recursive Function

```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0 or n == 1:
        return 1
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result
print(factorial(5)) 
print(factorial(0))
print(factorial(-3))
```

### Task 3: Lambda Function

```python
def lambda_fun(price):
    return price + (price * 0.18)
print(lambda_fun(100))

def create_lambda_fun(discount_percent):
    return lambda price: price - (price * (discount_percent / 100))
new_price = lambda_fun(100)
discounted_price = create_lambda_fun(10)
final_price = discounted_price(new_price)
print("discount on gst includes price", final_price,"gst includes price", new_price)
```

### Task 5: Using Filter

```python
prices_fil = [100,250,400,1200,50,2000,850]
prices_greather_than_500 = list(filter(lambda price: price > 500, prices_fil))
print("Prices greater than 500:", prices_greather_than_500)
prices_less_than_with_equal_to_500 = list(filter(lambda price: price <= 500, prices_fil))
print("Prices less than or equal to 500:", prices_less_than_with_equal_to_500)
```

### Task 6: Combined Utility Function

```python
def process_prices(prices):
    discounted_prices = list(map(lambda price: price - (price * 0.10), prices))
    filtered_discounted_prices = list(filter(lambda price: price > 300, discounted_prices))
    return discounted_prices, filtered_discounted_prices
prices = [100, 500, 900, 50,750]
discounted_prices, filtered_discounted_prices = process_prices(prices)
print("Discounted Prices:", discounted_prices)
print("Filtered Discounted Prices (greater than 300):", filtered_discounted_prices)
```

### Task 7: Mini Problem: Menu Using Function

```python
def add_price(prices_list, price):
    prices_list.append(price)
    return prices_list
def get_avarage_price(prices_list):
    if len(prices_list) == 0:
        return 0
    return sum(prices_list) / len(prices_list)
def get_max_price(prices_list):
    if len(prices_list) == 0:
        return None
    return max(prices_list)
#then create menu
prices_list = []
while True:
    print("\nMenu:")
    print("1. Add Price")
    print("2. Get Average Price")
    print("3. Get Maximum Price")
    print("4. Exit")
    choice = input("Enter your choice: ")
    print("Choice from user", choice)
    if choice == '1':
        price = float(input("Enter the price to add: "))
        add_price(prices_list, price)
        print(f"Price {price} added. Current Prices: {prices_list}")
    elif choice == '2':
        average_price = get_avarage_price(prices_list)
        print(f"Average Price: {average_price} of {prices_list} prices")
    elif choice == '3':
        max_price = get_max_price(prices_list)
        print(f"Maximum Price: {max_price} from {prices_list} prices")
    elif choice == '4':
        print("Exiting the program.")
        break
    else:
        print("Invalid choice. Please try again.")
```
