#### Task 1: Write Sales Record to a file


```python
sales = [1200,450,980,1500,3000]
with open('sales_data.txt', 'w') as file:
    for sale in sales:
        file.write(str(sale) + '\n')
    file.close()   
with open('sales_data.txt', 'r') as file:
    for line in file:
        print(line.strip()) 
    file.close()
# Extra Work
with open('sales_data_inline.txt', 'w') as file:
    for sale in sales:
        file.write(str(sale) + ',')
    file.close()
with open('sales_data_inline.txt', 'r') as file:
    data = file.read()
    print(data.strip(','))
    file.close()
```

### Task 2: Read File in Different Ways

```python
#sales_data.txt reading by read method
with open('sales_data.txt', 'r') as file:
    data = file.read()
    print(data)
    file.close()
print('---------------------------------------')
#sales_data.txt reading first line by readline method
with open('sales_data.txt', 'r') as file:
    first_line = file.readline()
    print(first_line.strip())
    file.close()
print('---------------------------------------')
#sales_data.txt reading all lines by readlines method
with open('sales_data.txt', 'r') as file:
    lines = file.readlines()
    list = []
    for line in lines:
        list.append(line.strip())
    print("Sales data List : ",list)
    file.close()
```

### Task 4: Generate Summary Report from a file

```python
with open('sales_data.txt', 'r') as file:
    list_of_sales = []
    for line in file:
        list_of_sales.append(int(line.strip()))
    total_sales = sum(list_of_sales)
    print("Total sales: ", total_sales)
    highest_sale = max(list_of_sales)
    print("Highest sale: ", highest_sale)
    lowest_sale = min(list_of_sales)
    print("Lowest sale: ", lowest_sale)
    avarage_sale = total_sales / len(list_of_sales)
    print("Avarage sale: ", avarage_sale)
    file.close()
```

### Task 6: Read File Safely (Error Handling inside File Handling only)

```python
# ask user to give the naname of file if exist then read the file and print the content otherwise print file not found
# with safe file handling using try except block
file_name = input("Enter the name of the file to read: ")
try:    
    with open(file_name, 'r') as file:
        content = file.read()
        print(content)
        file.close()
except FileNotFoundError:
    print("File not found. Please check the file name and try again.")

```

### Task 7: Mini Project - Export Discounted Price

```python
def is_number(text):
    try:
        float(text)
        return True
    except ValueError:
        return False
prices = {
    "Mouse" : 500,
    "Keyboard" : 800,
    "Monitor" : 7000,
    "Pen Drive" : 400,
    "Camera" : 5000
}
discounted_price = int(input("Enter the discount percentage: "))
print('---------------------------------------')
print("Generating discounted report with the given discount percentage: ", discounted_price)
print('---------------------------------------')
with open("discounted_report.txt","w") as file:
    file.write("Product Name - Original Price - Discounted Price\n")
    for product, price in prices.items():
        discounted_price_value = price - (price * discounted_price / 100)
        file.write(f"{product} - {price} - {discounted_price_value}\n")
    file.close()
print("Discounted report has been generated in discounted_report.txt")
print('---------------------------------------')
print("Printing summary of the report:")
print('---------------------------------------')
with open("discounted_report.txt","r") as file:
    total_items = 0
    avarage_discounted_price = 0
    list_of_discounted_prices = []
    for line in file:
        print(line.strip())
        total_items += 1
        line_parts = line.strip().split(" - ")
        if len(line_parts) == 3 and is_number(line_parts[2]):  # Check if the third part is a valid discounted price
            discounted_price_value = float(line_parts[2])
            list_of_discounted_prices.append(discounted_price_value)
    
    if list_of_discounted_prices:
        avarage_discounted_price = sum(list_of_discounted_prices) / len(list_of_discounted_prices)
    print('---------------------------------------')
    print("Printing total items and avarage discounted price: ")
    print('---------------------------------------')
    print("Total items: ", total_items - 1)  # Subtracting 1 for the header line
    print("Avarage discounted price: ", avarage_discounted_price)
    file.close()
```
