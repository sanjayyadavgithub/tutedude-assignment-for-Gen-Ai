## 1. What is bias in ML models

##### Bias is the error caused when a machine learning model is too simple to capture the important patterns in the data.
##### A model with high bias makes strong assumptions and performs poorly on both training and testing data.
##### Example: A very simple Decision Tree with max_depth=1 may have high bias

## 2. What is variance?

##### Variance is the error caused when a model is too sensitive to the training data.
##### A high-variance model learns not only the useful patterns but also noise and small details in the training data. It may perform very well on training data but poorly on new data.
##### Example: An unrestricted Decision Tree can have high variance.

## 3. How do bias and variance relate to underfitting and overfitting

| Situation        | Bias | Variance | Result               |
| ---------------- | ---- | -------- | -------------------- |
| **Underfitting** | High | Low      | Model is too simple  |
| **Good fit**     | Low  | Low      | Good generalization  |
| **Overfitting**  | Low  | High     | Model is too complex |


## 4. How can we reduce overfitting?

#### We can reduce overfitting by:
##### Using a simpler model
##### Limiting model complexity, such as restricting Decision Tree depth
##### Using more training data
##### Applying regularization
##### Removing unnecessary features
##### Using cross-validation
##### Avoiding excessive model tuning
