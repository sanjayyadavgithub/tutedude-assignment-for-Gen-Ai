```python
import pandas as pd
```

```python
df = pd.read_csv("student_data.csv")
df.head()
```

```python
G1 = df["G1"]
G1
```

```python
G2 = df['G2']
G2
```

```python
df["average_grade"] = (df["G1"] + df["G2"] + df["G3"]) / 3
```

```python
df["total_alcohol"] = df["Dalc"] + df["Walc"]
```

```python
def age_group(age):
    if age <= 16:
        return "Young"
    elif age <= 18:
        return "Teen"
    else:
        return "Adult"

```

```python
df["age_group"] = df["age"].apply(age_group)
df['age_group']
```

```python
def performance(score):
    if score >= 16:
        return "Excellent"
    elif score >= 12:
        return "Good"
    elif score >= 8:
        return "Average"
    else:
        return "Poor"
```

```python
df["performance"] = df["G3"].apply(performance)
df["performance"]
```
