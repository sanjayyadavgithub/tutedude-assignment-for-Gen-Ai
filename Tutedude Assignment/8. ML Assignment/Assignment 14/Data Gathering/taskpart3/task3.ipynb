```python
import sqlite3
import pandas as pd
```

```python
connectionDB = sqlite3.connect("sample.db")
cursor = connectionDB.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL
)
""")
```

```python
employees = [
    (1, "Alice", "HR", 50000),
    (2, "Bob", "IT", 65000),
    (3, "Charlie", "Finance", 55000),
    (4, "David", "Marketing", 48000),
    (5, "Emma", "IT", 70000)
]
```

```python

cursor.executemany(
    "INSERT OR REPLACE INTO employees VALUES (?, ?, ?, ?)",
    employees
)
```

```python
connectionDB.commit()
```

```python
df = pd.read_sql_query("SELECT * FROM employees", connectionDB)
```
