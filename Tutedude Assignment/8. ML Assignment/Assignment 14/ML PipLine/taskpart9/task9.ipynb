## Why are pipelines important in ML

#### Pipelines are important because they combine data preprocessing and machine learning models into a single workflow. They make the ML process easier, consistent, and reproducible. 
#### Raw Data → OneHotEncoding → MinMaxScaling → LinearRegression

# What problems do pipelines solve

#### Pipelines solve several common problems:
##### Data leakage: Preprocessing is fitted only on the training data.
##### Inconsistent preprocessing: The same transformations are automatically applied to training and test data.
##### Manual errors: No need to manually perform encoding and scaling every time.
##### Reproducibility: The same complete workflow can be reused.
##### Simpler code: Multiple preprocessing steps and the model are combined into one object.

# Manual preprocessing vs Pipeline-based preprocessing

###  Manual Preprocessing
##### Each step is performed separately
##### More code
##### Higher chance of mistakes
##### Easy to accidentally cause data leakage
##### Need to remember transformations for test data
##### Harder to maintain


### Pipeline-based Preprocessing
#####  All steps are connected together
#####  Less code
#####  Fewer preprocessing mistakes
#####  Helps prevent data leakage
#####  Automatically applies the same transformations
#####  Easier to maintain and reuse
