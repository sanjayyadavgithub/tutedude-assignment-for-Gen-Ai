```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_data.csv")
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="G1",
    y="G3",
    hue="sex"
)

plt.title("Scatter Plot: G1 vs G3")
plt.xlabel("G1")
plt.ylabel("G3")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.regplot(
    data=df,
    x="G1",
    y="G3",
    scatter_kws={"alpha":0.7}
)

plt.title("Regression Plot: G1 vs G3")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.lineplot(
    data=df.sort_values("G1"),
    x="G1",
    y="G3"
)

plt.title("Line Plot: G1 vs G3")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.kdeplot(
    data=df,
    x="G1",
    y="G3",
    fill=True
)

plt.title("Bivariate KDE Plot")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.kdeplot(
    data=df,
    x="G1",
    y="G3",
    fill=True
)

plt.title("Bivariate KDE Plot")

plt.show()
```

```python
plt.figure(figsize=(10,8))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.barplot(
    data=df,
    x="sex",
    y="G3"
)
plt.title("Average Final Grade by Gender")
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.boxplot(
    data=df,
    x="sex",
    y="G3"
)

plt.title("Box Plot of G3 by Gender")

plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.violinplot(
    data=df,
    x="sex",
    y="G3"
)

plt.title("Violin Plot")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.stripplot(
    data=df,
    x="sex",
    y="G3"
)

plt.title("Strip Plot")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.boxenplot(
    data=df,
    x="sex",
    y="G3"
)

plt.title("Boxen Plot")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="sex"
)

plt.title("Count Plot")

plt.show()
```
