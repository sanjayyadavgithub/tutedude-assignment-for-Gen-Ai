```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
df = pd.read_csv("student_data.csv")
df
```

```python
df.columns
```

```python
numerical_features = [
    'age',
    'Medu',
    'Fedu',
    'traveltime',
    'studytime',
    'failures',
    'famrel',
    'freetime',
    'goout',
    'Dalc',
    'Walc',
    'health',
    'absences',
    'G1',
    'G2',
    'G3'
]
```

```python
df[numerical_features].head()
```

```python
minmax_scaler = MinMaxScaler()
```

```python
df_minmax = df.copy()
```

```python
df_minmax[numerical_features] = minmax_scaler.fit_transform(
    df[numerical_features]
)
```

```python
df_minmax[numerical_features].head()
```

```python
df_minmax[numerical_features].min()
```

```python
df_minmax[numerical_features].max()
```

```python
standard_scaler = StandardScaler()
```

```python
df_standard = df.copy()
```

```python
df_standard
```

```python
df_standard[numerical_features] = standard_scaler.fit_transform(
    df[numerical_features]
)
```

```python
df_standard[numerical_features].head()
```

```python
comparison = pd.DataFrame({
    'Original G1': df['G1'].head().values,
    'MinMax G1': df_minmax['G1'].head().values,
    'Standard G1': df_standard['G1'].head().values
})
```
