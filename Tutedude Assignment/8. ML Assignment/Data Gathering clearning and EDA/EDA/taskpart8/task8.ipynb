```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_data.csv")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(
    data=df,
    x="G3",
    kde=True,
    bins=10,
    color="skyblue"
)

plt.title("Distribution of Final Grades (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.countplot(
    data=df,
    x="sex"
)

plt.title("Count of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data=df,
    x="G3"
)

plt.title("Boxplot of Final Grades (G3)")
plt.xlabel("Final Grade")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Histogram of G3")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.kdeplot(
    data=df,
    x="G3",
    fill=True
)

plt.title("KDE Plot of G3")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.rugplot(
    data=df,
    x="G3"
)

plt.title("Rug Plot of G3")

plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.stripplot(
    data=df,
    y="G3"
)

plt.title("Strip Plot of G3")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.boxenplot(
    data=df,
    x="G3"
)

plt.title("Boxen Plot of G3")

plt.show()
```
